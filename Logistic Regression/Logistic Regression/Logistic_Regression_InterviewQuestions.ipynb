```python
1. What is the difference between precision and recall?
Answer:
These concepts are fundamental for evaluating and validating your model.
To understand both, first recall the confusion matrix.
i)Precision:Of all the instances predicted as positive, how many were actually positive?
Precision=TP/TP+FP
where,True Positive (TP),False Positive (FP),False Negative (FN), True Negative (TN)
Key idea:
Focuses on false positives
High precision → few false alarms
Example use case:
Spam detection – You don’t want legitimate emails marked as spam.
ii)Recall (Sensitivity / True Positive Rate)
Recall answers the question:
Of all the actual positive instances, how many did the model correctly identify?
Recall=TP/TP+FN
​Key idea:
Focuses on false negatives
High recall → few missed positives
Example use case:
Disease detection – Missing a sick patient is costly.
iii)Precision vs Recall:Increasing precision often decreases recall and vice versa.
2. What is cross-validation, and why is it important in binary classification?
Answer:
What is Cross-Validation?
Cross-validation is a model evaluation technique where the dataset is split into multiple training and validation subsets to assess how well the model generalizes to unseen data.
1.K-Fold Cross-Validation (Most Common):
Split data into K equal folds
Train on K−1 folds
Validate on the remaining fold
Repeat K times
Average the performance metrics
Why Cross-Validation Is Important in Binary Classification
i)Better Generalization Estimate
Reduces dependency on a single train-test split
ii)Handles Class Imbalance More Reliably
Metrics like precision, recall, and F1-score are more stable
iii)Detects Overfitting
Large performance gap between folds signals poor generalization
iv)Efficient Use of Data
Every data point is used for both training and validation
2.Stratified Cross-Validation (Recommended)
For logistic regression in binary classification, stratified k-fold ensures each fold maintains the same class ratio as the original dataset
```
